```python
class Chocolate:
    def __init__(self, price: float, topping: str = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля.")

        self.price = price
        self.topping = topping

    def show_my_chocolate(self):
        if self.topping:
            print(f"Шоколадный батончик, добавка: {self.topping}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"





choc1 = Chocolate(price=120.5)
print(choc1)
choc1.show_my_chocolate()

print()


choc2 = Chocolate(price=99.9, topping="фундук")
print(choc2)
choc2.show_my_chocolate()

print()


choc3 = Chocolate(price=150.0, topping="карамель и морская соль")
print(choc3)
choc3.show_my_chocolate()

print()


try:
    choc_error = Chocolate(price=-10)
except ValueError as e:
    print(f"Ошибка: {e}")
```
