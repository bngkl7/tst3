```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('train.csv')

df.info()
```

```python
class_survival = df.groupby('Pclass')['Survived'].mean() * 100
class_survival
```

```python
def extract_name(full_name):
    parts = full_name.split(',')
    name_part = parts[1].strip()
    if '.' in name_part:
        return name_part.split()
    return name_part

df['First_Name'] = df['Name'].apply(extract_name)


male_names = df[df['Sex'] == 'male']['First_Name']
female_names = df[df['Sex'] == 'female']['First_Name']

most_common_male = male_names.value_counts().idxmax()
most_common_female = female_names.value_counts().idxmax()

print(f"\nСамое популярное мужское имя: {most_common_male}")
print(f"Самое популярное женское имя: {most_common_female}")
```

```python
for cls in [1, 2, 3]:
    class_df = df[df['Pclass'] == cls]
    male_names_cls = class_df[class_df['Sex'] == 'male']['First_Name']
    female_names_cls = class_df[class_df['Sex'] == 'female']['First_Name']

    print(f"\nКласс {cls}:")
    print(f"Самое популярное мужское имя: {male_names_cls.value_counts().idxmax()}")
    print(f"Самое популярное женское имя: {female_names_cls.value_counts().idxmax()}")
```

```python
print("\nПассажиры старше 44 лет:")
print(df[df['Age'] > 44])
```

```python
print("\nМужчины младше 44 лет:")
print(df[(df['Age'] < 44) & (df['Sex'] == 'male')])
```

```python
def count_cabins(df):
     cabin_groups = df.groupby('Cabin').size()
     return cabin_groups.value_counts()

print("\nКоличество n-местных кают:")
print(count_cabins(df))
```
